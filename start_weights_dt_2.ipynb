```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.pipeline import Pipeline
```

```python
df = pd.read_csv('cheating_detection_dataset_7.csv')
df
df = df.drop(columns = ['participant_id','competition_id'])
df
```

```python
df.describe()
```

```python
df

X = df.drop(columns = ['archetype'])
y = df['archetype']

from sklearn.model_selection import train_test_split


d = {x : y for x, y in zip(['hardworker', 'benchmark_tweaker', 'daily_sprinter',
       'consistent_late_breakthrough', 'low_interest_quitter',
       'consistent_ai_leap', 'ai_dump_sincere_start',
       'late_joiner_dangerous', 'pure_ai_quick_dump'], [0, 1, 2, 3, 4, 5, 6, 7, 8])}

display(d)
y = y.map(d)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, stratify = y, shuffle =True)

# train_df = df.copy()
# df_0 = train_df[train_df['Diabetes_binary'] == 0]
# df_1 = train_df[train_df['Diabetes_binary'] == 1]

# df_1 = df_1.sample(len(df_0), replace = True, random_state = 42)

# train_df = pd.concat([df_0, df_1])
# train_df = train_df.sample(frac= 1, random_state = 42)
# # train_df['row_id'] = train_df.index
# # train_df = train_df.drop(columns = ['row_id'])

# X_train = train_df.drop('Diabetes_binary', axis=1)
# y_train = train_df['Diabetes_binary']
# y_train = pd.get_dummies(y_train, dtype = int).values
```

```python
from sklearn.utils.class_weight import compute_sample_weight
```

```python
# display(X_train)
y_train = pd.DataFrame({
    'archetype' : y_train
})
train_df = pd.concat([X_train, y_train], axis = 1)
# df_0 = train_df[train_df['cheated'] == 0]
# df_1 = train_df[train_df['cheated'] == 1]
sample_weights = compute_sample_weight('balanced', y_train)

# df_1 = df_1.sample(len(df_0), replace = True, random_state = 42)

# train_df = pd.concat([df_0, df_1])
# train_df = train_df.sample(frac= 1, random_state = 42)


X_train = train_df.drop('archetype', axis=1)
y_train = train_df['archetype']




# y_train = pd.get_dummies(y_train, dtype = int).values
display(X_train, y_train)
display(y_train.value_counts())
```

```python
x1 = {'n_estimators': 957, 'max_depth': 10, 'learning_rate': 0.010489050210581588, 'subsample': 0.8816721785062149, 'colsample_bytree': 0.9393240336036071, 'min_child_weight': 1, 'reg_alpha': 0.029288256542840854, 'reg_lambda': 0.0033822170961102394, 'gamma': 0.5239400402156534}
xgb = XGBClassifier(**(x1))
xgb.fit(X_train, y_train, sample_weight=sample_weights)

```

```python
y_pred = xgb.predict(X_test)
f1_score_xgb = f1_score(y_test,y_pred, average='macro')
display(f1_score_xgb)
print(classification_report(y_test, y_pred))

```

```python
importance = pd.DataFrame({
    'Feature' : X_train.columns,
    'Importance' : xgb.feature_importances_
})

importance = importance.sort_values(by = 'Importance', ascending= False)
display(importance)
```
